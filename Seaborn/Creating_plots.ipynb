```python
tips = sns.load_dataset("tips")
print(type(tips))
print(tips.head)
```

```python
sns.set_theme(style="whitegrid")
sns.relplot(
    data = tips, 
    x = "total_bill",
    y = 'tip',
    hue = "smoker",
    size = "size",
    style = "smoker"
)
```

```python
x_values= [1,2,3,4,5,6,7,8]
values_square = [i ** 2 for i in x_values]
sns.set_theme(style="whitegrid")
sns.relplot( 
    kind = "line",
    x = x_values,
    y = values_square
)
```

```python
tips = sns.load_dataset("tips")
sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "time"
)
```

```python
sns.get_dataset_names()
flights_Data = sns.load_dataset("flights")
print(flights_Data.head)
sns.lineplot(
    data = flights_Data,
    x = "month",
    y = "passengers",
    hue = "year"
)
```

```python
# Categorical PLots - "Bar","Box"
print(tips.head())
sns.barplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```

```python
# print(tips.head())
sns.boxplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```

```python
# Distribution plots - Histogram

penguins = sns.load_dataset("penguins")
penguins.head()

sns.histplot(
    data = penguins,
    x = "body_mass_g",
    bins = 30
)
```

```python
# HeatMaps - PLot rectangular data as color-encoded matrix
flights = sns.load_dataset("flights")

flight_data = flights.pivot(index="month",columns="year",values="passengers")
sns.heatmap(
    flight_data,
    cmap = "coolwarm",
    annot = True,
    fmt = "d"
)
plt.title("Passengers Data")
plt.tight_layout()
```

```python
# Practive Seaborn
tips = sns.load_dataset("tips")
fig,ax = plt.subplots()
sns.lineplot(
    data = tips,
    x = "day",
    y = "total_bill",
    hue = "sex",
    marker="o",
    ax = ax
)

ax.set_title("Average Bill Value each day")
ax.set_xlabel("Days")
ax.set_ylabel("Total Bills Amount")
fig.tight_layout()
```
