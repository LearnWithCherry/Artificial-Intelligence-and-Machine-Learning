```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
data_set = sns.get_dataset_names()
data_set
```

```python
dots_data = sns.load_dataset("dots")
dots_data.head()
```

```python
sns.relplot(
    data=dots_data,
    x = "time",
    y="firing_rate",
    hue="coherence"
)
```

```python
tips = sns.load_dataset("tips")
sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    col="time",
    hue="smoker"
)
```

```python
planet_data = sns.load_dataset("planets")
pd.set_option("display.max_rows",None)
planet_data.head
sns.relplot(
    data = planet_data,
    x = "orbital_period",
    y = "distance",
    hue = "year",
    markers = "o"
)
plt.tight_layout()
plt.grid(True)
```
