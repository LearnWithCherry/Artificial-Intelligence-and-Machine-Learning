```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# sns.get_dataset_names()
data_set = sns.load_dataset("penguins")
print(data_set.head(10))
```

```python
sns.relplot(
    data = data_set,
    x = "flipper_length_mm",
    y = "body_mass_g",
    hue = "species"    
)
```

```python
tips = sns.load_dataset("tips")
print(tips.head(6))
sns.boxplot(
    data = tips,
    x = "total_bill",
    y = "day",
    hue = "sex"
)
```

```python
tips_scatter = sns.load_dataset("tips")
sns.scatterplot(
    data = tips_scatter,
    x = "total_bill",
    y = "tip",
    hue = "sex",
    style = "smoker",
)
plt.title("total_bill VS tip")
```

```python
tips = sns.load_dataset("tips")
tips_heatmap = tips.pivot_table(index="day", columns="time", values="tip", aggfunc="mean")
sns.heatmap(
    tips_heatmap, 
    cmap="coolwarm", 
    annot=True,
    fmt=".2f"
)
plt.title("Average Tip by Day and Time")
plt.show()
```
