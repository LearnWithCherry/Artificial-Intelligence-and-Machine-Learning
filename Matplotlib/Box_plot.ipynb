```python
# box plot - is a Statistical visualization that summarizes the distribution of a dataset using fice key values 
'''
1. Minimum
2. First Quartile (25% of data)
3. Median         (50% of data)
4. Third Quartile (75% of data)
5. Maximum
'''
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = [5,8,4,3,10,13,15,17,20]
plt.boxplot(data)
plt.grid()
```

```python
# Operation on Box plots
```

```python
# multiple data set on graph
data1 = np.random.normal(50,10,100)
data2 = np.random.normal(60,15,100)

plt.boxplot([data1,data2])
plt.grid()
```

```python
# horizontal box plot
data1 = np.random.normal(50,10,100)
data2 = np.random.normal(60,15,100)

plt.boxplot([data1,data2], vert = False)
plt.grid()
```

```python
# show mean
data1 = np.random.normal(50,10,100)
data2 = np.random.normal(60,15,100)

plt.boxplot([data1,data2],showmeans=True)
plt.grid()
```
