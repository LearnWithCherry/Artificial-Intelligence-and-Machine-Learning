```python
import matplotlib.pyplot as plt
```

```python
# data
people = ["People A", "People B", "People C", "People D", "People E", 
          "People F", "People G", "People H", "People I", "People J"]
age = [22,25,30,35,40,45,50,55,60,65]
blood_pressure = [110,115,120,122,125,130,135,123,145,150]
colors = ["Red" if x >= 135 else "green" for x in blood_pressure]


# Plotting data
plt.scatter(age,blood_pressure, s = blood_pressure, marker = "o", alpha = 1, cmap = "OrRd",c=blood_pressure)
plt.title("DIFFERENCE IN BLOOD PRESSURE AS AGE RISES")
plt.xlabel("AGE")
plt.ylabel("BLOOD PERSSURE")
plt.grid()
plt.colorbar(label = "Blood Pressure")

for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], blood_pressure[i]), xytext=(age[i], blood_pressure[i]))
```

```python
# Multiple data
cities = ["City A","City B","City C","City D","City E"]
winter_temp = [5,2,10,0,7] # in (C)
winter_humidity = [80,75,65,85,70] # in (%)
summer_temp = [25,30,28,35,27] # in (C)
summer_humidity = [60,50,55,45,65]# in (%)

plt.scatter(winter_temp,winter_humidity, label="Winter Data")
plt.scatter(summer_temp,summer_humidity, label="Summer Data")
plt.title("Summer VS Winter Data")
plt.xlabel("Temperature (C)")
plt.ylabel("Humidity (%)")
plt.legend()
```
