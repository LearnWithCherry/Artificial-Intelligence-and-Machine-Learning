```python
# Movies Data 
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008,2009,2010,2011,2012]
oscar_revenue = [1005,170,427,133, 232]
plt.title("Oscar Movies Revenue In Each year.")
plt.xlabel("Years")
plt.ylabel("Revenue In Million '$'")
plt.plot(years,oscar_revenue)
```

```python
non_oscar_movies = [
    "Slumdog Millionaire",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]
non_oscar_revenue = [378, 2788, 829,185, 275]

plt.title("Oscar Movie Revenue VS Non-Oscar Movie Revenue")
plt.plot(years,oscar_revenue,color = "red",marker = "o",linestyle = "-.",label = "Oscar Movie")
plt.plot(years,non_oscar_revenue,color = "green",marker = "o",linestyle = "-.",label = "Non-Oscar Movie")
plt.xlabel("Years")
plt.ylabel("Revenue In Million '$'")
plt.legend()

plt.style.available
# plt.style.use("dark_background")
plt.style.use("default")
plt.grid()
plt.show()
plt.savefig("file.png")
```
