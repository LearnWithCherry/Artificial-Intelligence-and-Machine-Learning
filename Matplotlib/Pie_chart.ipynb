```python
import matplotlib.pyplot as plt
```

```python
exp = ["Salary","Rent","Market","R&D","Miscellaneous"]
amount = [500,150,120,100,50]
plt.style.use("classic")
explode = [0,0,0,0.1,0]
plt.pie(amount,labels = exp, autopct="%1.1f%%",wedgeprops={
    "edgecolor":"black",
    "linewidth":1,
    "linestyle":"--",
    "fill" : True,
},  explode=explode,
    startangle = 90,
    shadow = True
)
```
