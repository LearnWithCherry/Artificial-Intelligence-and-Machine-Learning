```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# bar chart - A bar chart is a grapk that displays categorical data using rectangular bars
# each bar's height repersent values
```

```python
# DATA
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008,2009,2010,2011,2012]
oscar_revenue = [1005,170,427,133,232] # in M$

# PLOT DATA
plt.bar(years, oscar_revenue, color = "Red")
plt.title("Revenue Of Movies In Each Year")
plt.xlabel("Years")
plt.ylabel("Revenue In Million Dollars")
for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i]+20, oscar_revenue[i], ha="center")
plt.ylim(0,max(oscar_revenue)+100)
```

```python
# data 
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
non_oscar_movies = [
    "Slumdog Millionaire",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]
oscar_revenue = [1005,170,427,133, 232]
non_oscar_revenue = [378, 2788, 829,185, 275]
years = [2008,2009,2010,2011,2012]

# Chart 
x = np.arange(len(years))
width = 0.4  # Width of a single bar

plt.title("Oscar VS Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue In Million Dollars")

# The Logic: (x + offset)
plt.bar(x - width, oscar_revenue, width, label="Oscar Movies")
plt.bar(x, non_oscar_revenue, width, label="Non-Oscar Movies")
plt.xticks(x, years) # This ensures the year labels are centered
plt.legend()
plt.show()
```

```python
# Horizontal Chart
non_oscar_movies = [
    "Slumdog Millionaire",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]
non_oscar_revenue = [378, 2788, 829,185, 275]

plt.barh(non_oscar_movies,non_oscar_revenue )
plt.xlabel("Revenue.")
plt.ylabel("Movie Name.")
```
