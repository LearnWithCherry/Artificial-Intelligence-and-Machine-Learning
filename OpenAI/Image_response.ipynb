```python
from openai import OpenAI

API_openrouter = "API"
client = OpenAI(
    api_key=API_openrouter,
    base_url="https://openrouter.ai/api/v1"
)
```

```python
response = client.chat.completions.create(
    model="openrouter/free",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "text",
                    "text": "Analyze the car in the image and just tell me the name of the car!"
                },
                {
                    "type": "image_url",
                    "image_url": {
                        "url": "https://tse2.mm.bing.net/th/id/OIP.5lQLcYvWQtchQkhoQynJ-wHaEK?r=0&rs=1&pid=ImgDetMain&o=7&rm=3"
                    }
                }
            ]
        }
    ]
)

print(response.choices[0].message.content)
```
