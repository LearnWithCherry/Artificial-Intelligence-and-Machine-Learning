```python
from openai import OpenAI

API_openrouter = "API"
client = OpenAI(
    api_key=API_openrouter,
    base_url="https://openrouter.ai/api/v1"
)
```

## Email Summarizer

```python
class Email_assistant:
    def __init__(self):
        print("Hi, I am your personal AI Assistant.")

    def ans(self):
        email_text = input("Paste Your email: ")
        prompt = f"Summarize the Following email in 2-3 sentances: {email_text}" 
        response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages = [
            {"role" : "system","content" : "Act like an professional Summarizer."},
            {"role" : "user", "content": prompt}
        ],
            temperature=0.4,
            max_tokens=512
        )
        print("\n\nSummary: ",response.choices[0].message.content.strip())
```

```python
Email = Email_assistant()
```

```python
Email.ans()
```
