```python
from bs4 import BeautifulSoup
import pandas as pd
```

```python
with open(r"X://VScode//Artificial_intelligence_n_Machine_learning//Data_Collection//Web_Scrapping//scrapped_data.html",'r') as file:
    html_content = file.read()

soup = BeautifulSoup(html_content, "lxml")
```

```python
soup.find("h3")
```

```python
all_h3 = soup.find_all("h3")
all_countries = []
for h3 in all_h3:
    # print(h3)
    name = h3.get_text(strip=1)
    population = (h3.find_next("div").select("span.country-population")[0].get_text(strip=1))
    # print(h3.find_next("div").select_one("span.country-population").get_text(strip=0))
    # print(h3.attrs)

    all_countries.append([name, population])
    # all_countries.append([f"Country name: {name}. Population: {population}"])
```

```python
df = pd.DataFrame(all_countries, columns = ["Name", "Population"])
df
```

```python
df.to_csv("Cleaned_data.csv",index=False)
```
