```python
import requests
import pandas as pd 
```

```python
URL = "https://stephen-king-api.onrender.com/api/books"
response = requests.get(URL)
```

```python
if response.status_code == 200:
    with open("X://VScode//Artificial_intelligence_n_Machine_learning//Data Collection//Web_Scrapping//scrapped_data.html","w") as file:
        file.write(response.text)

json_data = response.json()
df = pd.json_normalize(json_data["data"])
df = df[["id","Title","Year","ISBN"]]
df
```

```python
# 1. Use the specific data endpoint
cat_url = "https://cat-fact.herokuapp.com/facts"

response = requests.get(cat_url)

# 2. Safety Check: Always check if the request was successful
if response.status_code == 200:
    try:
        data = response.json()
        df = pd.json_normalize(data)
        print(df.head())
    except requests.exceptions.JSONDecodeError:
        print("The server didn't return JSON. It returned:")
        print(response.text[:100]) # Print first 100 chars to see what it is
else:
    print(f"Error: Server returned status {response.status_code}")
```
