```python
# Problem Statement - From Quotes Extract all Quotes & auther from all the pages with life tag.
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd 
```

```python
# Collecting Data from Website
page_count = 1
while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    resouce = requests.get(URL)
    soup = BeautifulSoup(resouce.text, "lxml")
    quotes = soup.select("div.quote")
    if not quotes:
        print("No valid Quotes.")
        break
    with open(f"X://VScode//Artificial_intelligence_n_Machine_learning//Data_Collection//Web_Scrapping//Quotes{page_count}.html","w",encoding = "utf-8") as file:
        file.write(resouce.text)
        print(f"Downloaded data from {page_count}")

    page_count = page_count + 1
```

```python
# Extracting useful data from the file (Working with file "1")
with open(f"X://VScode//Artificial_intelligence_n_Machine_learning//Data_Collection//Web_Scrapping//Quotes1.html","r",encoding= "utf-8") as file:
    data = file.read()
soup = BeautifulSoup(data,"lxml")
# print(quotes)
```

```python
quotes = soup.select("div.quote")
life_quotes = []
for quote in quotes:
    all_tags = []
    for tag in quote.select(".tags .tag"):
        all_tags.append(tag.get_text())
        # print(tag.get_text())
    if "life" in all_tags:
        text = quote.select_one("span.text").get_text()
        author = quote.select_one("small.author").get_text()
        life_quotes.append([text, author])
```
