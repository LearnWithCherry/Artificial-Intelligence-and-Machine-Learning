```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.metrics import mean_absolute_percentage_error
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import root_mean_squared_error
from sklearn.metrics import accuracy_score
import numpy as np
```

```python
house_price_data = pd.read_csv("X:\\Artificial Intelligence Course\\Supervised Learning\\Linear Regression\\HousePricePrediction.csv")
house_price_data
```

```python
house_price_data.drop(['Id'],axis=1,inplace=True)
house_price_data = house_price_data.dropna(subset=["SalePrice"])
```

```python
categorical_cols = ["MSZoning", "LotConfig", "BldgType", "Exterior1st"]
house_price_data = pd.get_dummies(house_price_data, columns=categorical_cols, drop_first=False, dtype=int)
```

```python
house_price_data = house_price_data.dropna()
house_price_data.head(10)
```

```python
house_price_data.isnull().sum().head()
```

```python
X = house_price_data.drop(["SalePrice"],axis=1)
y = house_price_data["SalePrice"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.24, random_state=0)
```

```python
X_train.head()
scaler = StandardScaler()

# Get list of numerical columns (excluding your encoded 0/1 columns)
num_cols = ["MSSubClass", "LotArea", "OverallCond", "YearBuilt", "YearRemodAdd", "BsmtFinSF2", "TotalBsmtSF"]

# Fit on training data and transform both sets
X_train[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test[num_cols] = scaler.transform(X_test[num_cols])
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_series = pd.Series(y_pred)
print(f"Y predicted value: {y_series.sample(10)}")
```

```python
print(f"R2 score: {r2_score(y_test, y_pred):.4f}")
print(f"Mean absolute error: {mean_absolute_error(y_test, y_pred):.4f}")
print(f"Root mean squared error: {root_mean_squared_error(y_test, y_pred):.4f}")
print(f"Mean absolute percentage error :{mean_absolute_percentage_error(y_test,y_pred):.4f}")
# print(f"Accuracy Score: {accuracy_score(np.array([[0, 1], [1, 1]]), np.ones((2, 2))):.4f}")
```
