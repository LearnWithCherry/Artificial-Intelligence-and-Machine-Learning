```python
X = employee_data.drop("Employee_Turnover",axis=1)
y = employee_data["Employee_Turnover"]
```

```python
X = pd.DataFrame(X)
X.head(2)
```

```python
y = pd.DataFrame(y)
y.head(2)
```

```python
X_train, X_test, y_train, y_text = train_test_split(
    X, y, test_size=0.25, random_state=42)
```

```python
# 1. Logistic Regression WITH L2 Regularization (Default)
log_reg_l2 = LogisticRegression(penalty='l2', C=1.0, max_iter=1000, random_state=42)
log_reg_l2.fit(X_train, y_train.values.ravel())
y_pred_l2 = log_reg_l2.predict(X_test)
y_proba_l2 = log_reg_l2.predict_proba(X_test)[:, 1]

# 2. Logistic Regression WITHOUT Regularization
log_reg_none = LogisticRegression(penalty=None, max_iter=1000, random_state=42)
log_reg_none.fit(X_train, y_train.values.ravel())
y_pred_none = log_reg_none.predict(X_test)
y_proba_none = log_reg_none.predict_proba(X_test)[:, 1]
```
