```python
import numpy as np

class LinearRegressionOLS:
    def __init__(self):
        self.bias = None
        self.weight = None

    def fit(self, X, y):
        m, n = X.shape
        # Add column of ones for the bias term
        X_b = np.c_[np.ones((m, 1)), X]
        # The Normal Equation: theta = (X^T * X)^-1 * X^T * y
        theta = np.dot(np.linalg.inv(np.dot(X_b.T, X_b)), np.dot(X_b.T, y))
        
        self.bias = theta[0]
        self.weight = theta[1:]
        
    def predict(self, X):
        # Y = bias + X * weights
        y_pred = self.bias + np.dot(X, self.weight)
        return y_pred

# Test the model
X = np.array([[1],[2],[3],[4],[5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegressionOLS()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

print(model.bias)
print(model.weight)
```
