```python
# Merging andd joining Data

df_costomer = pd.DataFrame({
    "Customer_ID" : [1,2,3,4,5],
    "Name" : ["Aman","Aisha","Ruhi","Ankita","Rajat"]
})

df_order = pd.DataFrame({
    "Order_ID" : [101,102,103,104,105],
    "Customer_ID"  :[1,2,4,5,6],
    "Amount" : [23,345,124,4566,133]
})
```

```python
pd.merge(df_costomer, df_order, on="Customer_ID") # inner join
```

```python
pd.merge(df_costomer, df_order, on="Customer_ID",how="left") # left join
```

```python
pd.merge(df_costomer, df_order, on="Customer_ID",how="right") # right join
```

```python
pd.merge(df_costomer, df_order, on="Customer_ID",how="outer") # ovter join
```
