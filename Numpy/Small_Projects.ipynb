```python
# ----------------------PROJECT 1----------------------
# ----------------------DATA PART----------------------
# Student names
students = np.array(['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'])

# Subject names
subjects = np.array(['Math', 'English', 'Science', 'History'])

# 2D array: 10 students × 4 subjects (scores 0-100)
# Each row = one student, Each column = one subject
grades = np.array([
    [85, 78, 92, 88],      # Alice
    [76, 82, 79, 85],      # Bob
    [92, 88, 95, 91],      # Charlie
    [88, 91, 67, 89],      # Diana
    [78, 75, 80, 77],      # Eve
    ])

# print("Students:", students)
# print("Subjects:", subjects)
# print("\nGrades Array (5 students × 4 subjects):")
# print(grades)
# ----------------------CALCULATION PART----------------------

student_average = grades.mean(axis = 1)
subject_average = grades.mean(axis = 0)
# min_avgerage = grades.mean(axis = 0)

print("Average of Each Student.")
for i in range(len(students)):
    print(f"{students[i]} Average is: {student_average[i]}")
    
max_average_class = np.max(student_average)
max_average_subject = np.max(subject_average)
min_average_subject = np.min(subject_average)
print(50*"-")
print("Calculaing Average.")
print(f"Highest Average in the Class: {max_average_class}")
print(f"Highest Average in the subject: {max_average_subject}")
print(f"Lowest Average in the subject: {min_average_subject}")
print(50*"-")

student_passed_or_not = np.all(grades >= 70, axis = 1)
print("List of passed and Failed Students.")
for i in range(len(students)):
    if student_passed_or_not[i]:
        print(f"{students[i]} is Passed with {student_average[i]}.")
    else:
        print(f"{students[i]} is Failed with {student_average[i]}.")

    
```

```python
# Task 2: Broadcasting - Temperature Data

Temperature_Readings = np.array([[25.5, 26.0, 25.8, 26.2],
                                 [30.1, 30.5, 30.2, 30.8],
                                 [22.3, 22.5, 22.1,22.9]])
baseline_temperature = np.array([25.0, 30.0, 22.0]).reshape(3,1)
Difference_Temperature =  Temperature_Readings - baseline_temperature
print(f"Original Temperature Reading: \n{Temperature_Readings}")
print(f"BaseLine Temperature Reading: \n{baseline_temperature}")
print(f"Temperature Difference after baseline Deduction: \n{Difference_Temperature}")
print(f"Shape of the result Data: {Difference_Temperature.shape}")
```

```python
# Task 3: Advanced Boolean Indexing - Filter Abnormal Readings
sensor_data = np.array([45, 67, 23, 89, 12, 34, 56, 78, 90, 11, 22, 33, 95, 44, 76, 88])
# print(sensor_data.size)
count_abnormal_reading = sensor_data[sensor_data > 75]
Readings_in_Normal = sensor_data[(sensor_data <= 70) & (sensor_data >= 30)] 
conditional_replacement = np.where(sensor_data > 85, 85, sensor_data)
print(f"Original Data: {sensor_data}")
print(f"Filtered Data: {count_abnormal_reading}")
print(f"Count of Abnormal readings: {count_abnormal_reading.size}")
print(f"Readings between 30 and 70: {Readings_in_Normal}")
print(f"Count of Normal readings: {Readings_in_Normal.size}")
print(f"After clipping values > 85 to 85: {conditional_replacement}")
```
