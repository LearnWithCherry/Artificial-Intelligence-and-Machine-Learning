```python
import numpy as np

# dataset: rows = students, columns = [math_score, physics_score]
X = np.array([[80, 90],
              [70, 85],
              [60, 75],
              [90, 95]])

mean = np.mean(X, axis=0)
std = np.std(X, axis=0)

X_normalized = (X - mean) / std

print("Original Data:\n", X)
print("Mean:", mean)
print("Standard Deviation:", std)
print("Normalized Data:\n", X_normalized)
```

```python
import numpy as np

# data
X = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])

# compute slope and intercept manually
mean_x = np.mean(X)
mean_y = np.mean(y)

numerator = np.sum((X - mean_x) * (y - mean_y))
denominator = np.sum((X - mean_x)**2)

m = numerator / denominator
b = mean_y - m * mean_x

print("Slope:", m)
print("Intercept:", b)
```
