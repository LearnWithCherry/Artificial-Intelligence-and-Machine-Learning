```python
# Dictionary (Most Important Structure)
# A dictionary is:
# Mutable
# Key-value pairs
# Unordered (insertion order preserved in modern Python)
# Keys must be immutable
```

```python
d = {
    "name"  :   "Cherry",
    "age"   :   20
#   keys    :   values
}
d
```

```python
# accessing value

d["name"] # If key doesn’t exist → KeyError.

d.get("name") 
d.get("grade", "Not Found") # return not found when key is not present
```

```python
# adding
d["city"] = "Delhi"
d
```

```python
# Updating
d["age"] = 19
d
```

```python
# Removing 
d.pop("Age",None)
d
```

```python
del d["name"]
d
```

```python
# 🔹 Important Dictionary Methods
```

```python
d.keys()
```

```python
d.values()
```

```python
d.items()
```

```python
# Iterating dictionary 

for key in d:
    print(key)

for key, value in d.items():
    print(key, value)
```

```python
# Dictionary Comprehension
square = {x: x*x for x in range(5)}
square
```
