```python
# Lambda Function - (lambda parameters : expression)
square = lambda x: x*x

print(square(5))
```

```python
# map() function 

nums = [1,2,3,4,5]
square = list(map(lambda x:x*x, nums))
print(square)
```

```python
nums = [1,2,3,4,5]
evens = list(filter(lambda x:x*x%2==0,nums))
evens
```
