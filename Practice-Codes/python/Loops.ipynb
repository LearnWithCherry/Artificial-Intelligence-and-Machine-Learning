```python
# prime number 
for num in range(1, 21):
    if num < 2:
        print(num, "-> Not Prime")
    else:
        is_prime = True
        
        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break
        
        if is_prime:
            print(num, "-> Prime")
        else:
            print(num, "-> Not Prime")
```

```python
# loop with enum, zip and else
```

```python
# normal iteration using loop
fruits = ["Apple","Mango","Guava","Banana","Kivi"]
for fruit in fruits:
    print(fruit)
```

```python
fruits = ["Apple","Mango","Guava","Banana","Kivi"]
for i in range(len(fruits)):
    print(i,fruits[i])
```

```python
# using Enumerate function 
for index, value in enumerate(fruits):
    print(index, value)
```

```python
# using ZIP function 
```

```python
names = ["Rajat","Priya","Ansh","Ruhi"]
marks = [12,23,45,56]
for name, mark in zip(names, marks):
    print(name, "->",mark)
```

```python
names = ["A", "B", "C"]
scores = [50, 60, 70]

for index, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"{index}. {name} scored {score}")
```
