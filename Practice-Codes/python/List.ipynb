```python
# accessing elements using index

L = [1,2,3,4,5,6,7,8,9]
L[2] # Access elements using index
L[-1] # access last elements
L[0:] # access all the elements 
L[0:2] # Slicing values (slicing create a new list)
L[0] = 100 # changing values in an list because it is mutable
```

```python
# append() add one element at the back of the list, Modifies the list 
list1 = [1,2,3,4]
list1.append(5)
list1
nums = [1,2,3,4]
print(nums.append(5)) # return none when prints
```

```python
# extend() for multiples elements

list2 = [1,2,3]
list2.extend([4,5,6,7])
list2
```

```python
# insert() insert at a specific index

list3 = [1,3,4,5]
list3.insert(1,2) # index , value
list3
```

```python
# remove remove first matching value

list4 = [1,2,3,4,5]
list4.remove(1) # values not index
list4
# error if not found
```

```python
# pop removes by index

list5 = [1,2,3,4,5,100]
list5.pop() # last
list5.pop(0) # first
list5.pop(2) # index
list5
```
