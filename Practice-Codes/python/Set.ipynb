```python
# set operator 

a = {1,2,3}
b = {3,4,5}

a | b # Union remove repeated value from set when combine into one 

a & b # Intersection return repeating value's

a - b # Difference return elements that are presnt in one but not in other set
b - a # Difference return elements that are presnt in one but not in other set

a ^ b # does not print repeating value's
```

```python
# program

new_set1 = {1,2,3,4}
new_set2 = {4,5,6,7}

new_set2.add(8)

new_set1 | new_set2 #union
new_set1 & new_set2 #intersection

new_set3 = new_set1 | new_set2
new_set3.clear()
new_set3
```
