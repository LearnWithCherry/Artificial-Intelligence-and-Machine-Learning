```python
class Student:
    def __init__(self, name="Not Defined", language="Not Defined", year="N/A", course="Not Defined"):
        self.name = name
        self.language = language
        self.year = year
        self.course = course
        
    def get_detail(self):
        print(f"Name of the Student is {self.name}\n"
             f"Student is in {self.year}\n"
             f"Student is Studying {self.language} language\n"
             f"Student Course is {self.course}")
    
    def set_detail(self):
        self.name = input("Enter name: ")
        self.language = input("Enter language: ")
        self.year = input("Enter year: ")
        self.course = input("Enter course: ")

    
std1 = Student()
std1.set_detail()
std1.get_detail()
```

```python
# method
class laptop:
    storage_type = "SSD"

    def __init__(self, RAM, storage, CPU):
        self.RAM = RAM
        self.storage = storage
        self.CPU = CPU
    @classmethod
    def get_storage_type(cls):
        return f"Storage type: {cls.storage_type}"
    def get_info(self):
         return f"Laptop has {self.RAM} RAM\nLaptop has {self.CPU} CPU\nLaptop has {self.storage} Storage\nStorage type is {self.storage_type}"

l1 = laptop("16GB","1TB","Ryzen")
l2 = laptop("24GB","2TB","intel")
l3 = laptop("32GB","4TB","Threadripper")

print(l3.get_info())
l1.get_storage_type()
```
