```python
#code for rmse
import numpy as np

y = np.array([3,5,7,9,11])
yp = np.array([2.5,5.5,6,9.5,10])
MSE = np.mean((y-yp)**2)
RMSE = np.sqrt(MSE)
print("MSE:",MSE)
print("RMSE:",RMSE)
print(f"RMSE:{RMSE:.4f}")
```
