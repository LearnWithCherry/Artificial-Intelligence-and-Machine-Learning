  Compute L1, L2, and L∞ norms, inner products, and vector projections to analyze feature similarity.

```python
# Import NumPy library for numerical computations
import numpy as np

# Define two feature vectors (example: two data samples)
x = np.array([3, 4])
y = np.array([1, 2])

# -----------------------------
# 1. Compute Vector Norms
# -----------------------------

# L1 Norm (Manhattan Distance): sum of absolute values
l1_norm = np.linalg.norm(x, ord=1)

# L2 Norm (Euclidean Distance): square root of sum of squares
l2_norm = np.linalg.norm(x, ord=2)

# L-infinity Norm (Maximum Norm): maximum absolute value
linf_norm = np.linalg.norm(x, ord=np.inf)

print("L1 Norm of x:", l1_norm)
print("L2 Norm of x:", l2_norm)
print("L-infinity Norm of x:", linf_norm)

# -----------------------------
# 2. Inner Product (Dot Product)
# -----------------------------

# Measures similarity between two vectors
inner_product = np.dot(x, y)

print("\nInner Product of x and y:", inner_product)

# -----------------------------
# 3. Vector Projection
# -----------------------------

# Projection of vector x onto vector y
# Formula: proj_y(x) = (x · y / y · y) * y
projection = (np.dot(x, y) / np.dot(y, y)) * y

print("\nProjection of x onto y:", projection)

# -----------------------------
# Interpretation
# -----------------------------
# Higher inner product → higher similarity between features
# Projection shows how much x aligns with y
```

L1 norm → feature sparsity

L2 norm → magnitude of feature vector

L∞ norm → dominant feature

Inner product → similarity between data points

Projection → component of one feature along another
