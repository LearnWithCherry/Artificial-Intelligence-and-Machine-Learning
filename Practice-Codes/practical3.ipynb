Practical 3: Genetic Algorithm for Continuous Function Optimization

Aim:- Optimize a continuous objective function using a genetic algorithm and observe convergence

```python
import numpy as np

def fitness(x):
    return -(x**2 + 5*np.sin(x))   # maximize fitness

pop = np.random.uniform(-5, 5, 20)

for _ in range(50):
    fit = fitness(pop)
    parents = pop[np.argsort(fit)[-10:]]
    children = parents + np.random.normal(0, 0.2, 10)
    pop = np.concatenate((parents, children))

best = pop[np.argmax(fitness(pop))]
print("Best solution:", best)
```

KEY STEPS:
1. intialize population with real values
2. evaluate fitness function
3. select best individuals (parents)
4. Apply mutation to generate offspring.
5. Replace population and repeat


What students observe:
✔ Fitness improves over generations
✔ Convergence toward optimal value
