```python
size = 1_000_000

py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"Python list time {end-start} second")

np_arr = np.array(py_list)
start = time.time()
sq_array = np_arr ** 2
end = time.time()
print(f"NUmpy array time {end-start} second")
```

```python
import sys
print(f"Python list size = {sys.getsizeof(py_list) * len(py_list)} Bytes")
print(f"Numpy list size = {np_arr.nbytes} Bytes")
```
