Benchmark Functions Used

Sphere Function (Convex, unimodal)

Rastrigin Function (Non-convex, multimodal)

```python
import numpy as np


def sphere(x):
    return np.sum(x**2)

def rastrigin(x):
    n = len(x)
    return 10*n + np.sum(x**2 - 10*np.cos(2*np.pi*x))

# Random solution vector (2D)
x = np.array([1.5, -2.0])

# Evaluate functions
print("Input vector:", x)
print("Sphere function value:", sphere(x))
print("Rastrigin function value:", rastrigin(x))
```

Sphere Function

Simple and smooth

Only one global minimum

Used to test basic convergence of optimizers

👉 Easy for gradient-based methods

Rastrigin Function

Has many local minima

Difficult search space

Used to test global search ability

👉 Suitable for GA, PSO, and other metaheuristics

Why Benchmark Functions Are Used

To compare optimization algorithms

To test convergence speed

To check ability to escape local minima

Conclusion

Sphere function is easy to optimize

Rastrigin function is challenging

Good optimizers perform well on both
